```python
import pandas as pd
```

```python
df = pd.read_csv("india_statewise_vaccine_template.csv")
```

```python
print("Rows and Columns:", df.shape)
```

```python
print(df.columns)
```

```python
df.describe()
```

```python
print(df.isnull().sum())
```

```python
first_dose = df[['State', 'First Dose Administered']]
print(first_dose)
```

```python
second_dose = df[['State', 'Second Dose Administered']]
print(second_dose)
```

```python
male_total = df['Male(Individuals Vaccinated)'].sum()
print("Total Male Vaccinated:", male_total)
```

```python
female_total = df['Female(Individuals Vaccinated)'].sum()
print("Total Female Vaccinated:", female_total)
```

```python
top10_first = df.sort_values(by='First Dose Administered', ascending=False).head(10)
print(top10_first[['State', 'First Dose Administered']])
```

```python
top10_second = df.sort_values(by='Second Dose Administered', ascending=False).head(10)
print(top10_second[['State', 'Second Dose Administered']])
```

```python
total_first = df['First Dose Administered'].sum()
total_second = df['Second Dose Administered'].sum()

print("Total First Dose in India:", total_first)
print("Total Second Dose in India:", total_second)
```

```python
df['Vaccination Percentage'] = (df['Second Dose Administered'] / df['First Dose Administered']) * 100
print(df[['State', 'Vaccination Percentage']])
```

```python
highest_male = df.loc[df['Male(Individuals Vaccinated)'].idxmax()]
print(highest_male[['State', 'Male(Individuals Vaccinated)']])
```

```python
highest_female = df.loc[df['Female(Individuals Vaccinated)'].idxmax()]
print(highest_female[['State', 'Female(Individuals Vaccinated)']])
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(14,7))
plt.bar(df['State'], df['First Dose Administered'])
plt.xticks(rotation=90)
plt.title("State-wise First Dose Vaccination")
plt.xlabel("State")
plt.ylabel("First Dose")
plt.show()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(14,7))
plt.bar(df['State'], df['Second Dose Administered'])
plt.xticks(rotation=90)
plt.title("State-wise Second Dose Vaccination")
plt.xlabel("State")
plt.ylabel("Second Dose")
plt.show()
```

```python
male_total = df['Male(Individuals Vaccinated)'].sum()
female_total = df['Female(Individuals Vaccinated)'].sum()

plt.figure(figsize=(6,6))
plt.pie([male_total, female_total],
        labels=['Male', 'Female'],
        autopct='%1.1f%%')
plt.title("Male vs Female Vaccination")
plt.show()
```

```python
print(df[['First Dose Administered',
          'Second Dose Administered',
          'Male(Individuals Vaccinated)',
          'Female(Individuals Vaccinated)']].corr())
```

```python
df['Vaccination Percentage'] = (df['Second Dose Administered'] / df['First Dose Administered']) * 100

plt.figure(figsize=(14,6))
plt.plot(df['State'], df['Vaccination Percentage'], marker='o')
plt.xticks(rotation=90)
plt.title("State-wise Vaccination Percentage")
plt.xlabel("State")
plt.ylabel("Percentage")
plt.grid(True)
plt.show()
```
